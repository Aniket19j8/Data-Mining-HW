```python
# running this befor modeling cell
train_raw = pd.read_csv("train.csv")
test_raw  = pd.read_csv("test.csv")
combine = [train_raw.copy(), test_raw.copy()]
```

```python
import numpy as np, pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.impute import SimpleImputer
from sklearn.model_selection import StratifiedKFold, cross_val_score

from sklearn.linear_model import LogisticRegression, SGDClassifier, Perceptron
from sklearn.svm import SVC, LinearSVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

train_df = combine[0].copy()
y = train_df["Survived"].values
X = train_df.drop(columns=["Survived"]).copy()

# feature tweaks
X["Title"] = X["Name"].str.extract(r",\s*([^\.]+)\.", expand=False)
X["Title"] = X["Title"].replace({"Mlle":"Miss","Ms":"Miss","Mme":"Mrs",
                                 "Lady":"Royalty","Countess":"Royalty","Dona":"Royalty",
                                 "Sir":"Royalty","Don":"Royalty","Jonkheer":"Royalty",
                                 "Capt":"Officer","Col":"Officer","Major":"Officer","Dr":"Officer","Rev":"Officer"})

X["FamilySize"] = X["SibSp"].fillna(0) + X["Parch"].fillna(0) + 1
X["IsAlone"] = (X["FamilySize"]==1).astype(int)

ticket_counts = X["Ticket"].value_counts()
X["TicketGroupSize"] = X["Ticket"].map(ticket_counts).fillna(1)

X["CabinDeck"] = X["Cabin"].astype(str).str[0].replace({"n":np.nan})
X["AgeClass"] = X["Age"] * X["Pclass"]
X["FarePerPerson"] = X["Fare"] / X["FamilySize"].replace(0,np.nan)

#preprocessing
categorical = [c for c in X.columns if X[c].dtype=="object"]
numeric = [c for c in X.columns if c not in categorical]

def safe_log(arr): return np.log1p(np.clip(arr,0,None))

log_cols = [c for c in ["Fare","FarePerPerson"] if c in numeric]
other_num = [c for c in numeric if c not in log_cols]

num_tf = ColumnTransformer([
    ("log_cols", Pipeline([
        ("imp", SimpleImputer(strategy="median")),
        ("log", FunctionTransformer(safe_log, validate=False)),
        ("scaler", StandardScaler())
    ]), log_cols),
    ("other_num", Pipeline([
        ("imp", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), other_num)
], remainder="drop")

try:
    cat_tf = Pipeline([
        ("imp", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
    ])
except TypeError:
    cat_tf = Pipeline([
        ("imp", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore", sparse=False))
    ])

prep = ColumnTransformer([
    ("num", num_tf, numeric),
    ("cat", cat_tf, categorical)
])

# models:
models = {
    "SVM": SVC(kernel="rbf", C=1.5, gamma="scale"),
    "KNN": KNeighborsClassifier(n_neighbors=9, weights="distance"),
    "LogReg": LogisticRegression(max_iter=1000),
    "RandomForest": RandomForestClassifier(n_estimators=300, max_depth=None, min_samples_split=2, random_state=42),
    "NaiveBayes": GaussianNB(),
    "Perceptron": Perceptron(max_iter=1000, penalty="l2", alpha=1e-4, random_state=42),
    "SGD": SGDClassifier(loss="log_loss", alpha=1e-4, early_stopping=True, random_state=42),
    "LinearSVC": LinearSVC(C=1.0, max_iter=5000, dual="auto", random_state=42),
    "DecisionTree": DecisionTreeClassifier(max_depth=6, min_samples_leaf=12, random_state=42)
}

# CV looping
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
results = []

for name, clf in models.items():
    pipe = Pipeline([("prep", prep), ("clf", clf)])
    scores = cross_val_score(pipe, X, y, cv=skf, scoring="accuracy")
    results.append((name, scores.mean(), scores.std()))

results = sorted(results, key=lambda t:t[1], reverse=True)
for n,m,s in results:
    print(f"{n:15s}: {m:.4f} ± {s:.4f}")
```
